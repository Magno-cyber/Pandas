```python
import pandas as pd
import numpy as np

# Primeiro parametro: O encoding='utf-8' é necessário para lidar com caracteres especiais / Segundo parametro: O sep =';' é usado para definir o separador do arquivo CSV, que neste caso é ponto e vírgula /  Terceiro parametro: O low_memory=False é usado para evitar o aviso de memória ao ler arquivos grandes / Quarto parametro: O decimal=',' é usado para definir o separador decimal, que neste caso é vírgula

df_vendas = pd.read_csv('vendas_tech.csv', low_memory=False) 
display(df_vendas)
df_gerentes = pd.read_excel('gerentes_lojas.xlsx')
display(df_gerentes)
```

```python
# Tratamento de dados

#display(df_vendas["Loja"]) # Usado para selecionar uma coluna do DataFrame df_vendas
#display(df_vendas[["Cliente", "Produto", "Preco_Unitario"]]) # Usado para selecionar múltiplas colunas do DataFrame df_vendas

# ----------- Colunas-------------

df_analise = df_vendas.drop(columns=["Data_Base"]) # Usado para remover a coluna "Data_Base" do DataFrame df_vendas

# ------------Nulos---------------

#df_analise = df_analise.dropna() # Usado para remover linhas com valores nulos do DataFrame df_analise
df_analise["Loja"] = df_analise["Loja"].fillna("Online") # Usado para preencher valores nulos na coluna "Loja" do DataFrame df_analise com o valor "Loja não informada"

# --------Tipos de dados----------

df_analise['Data'] = pd.to_datetime(df_analise['Data'], format='%Y-%m-%d') # Usado para converter a coluna "Data" do DataFrame df_analise para o tipo datetime

# --------Padronização de dados----------

df_analise["Loja"] = df_analise["Loja"].str.strip()  # Remove espaços em branco no início e no final dos valores da coluna "Loja" do DataFrame df_analise
df_analise["Loja"] = df_analise["Loja"].str.title()  # Usada para padronizar os valores da coluna "Loja" do DataFrame df_analise para o formato de título (primeira letra maiúscula e as demais minúsculas)

# ----------Duplicados------------

df_analise = df_analise.drop_duplicates(subset=["ID_Pedido"]) # Usado para remover linhas duplicadas do DataFrame df_analise

display(df_analise)
```
