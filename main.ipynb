```python
# Tratamento de dados
#  
#display(tabela1['Loja'])
#display(tabela1[['Loja', 'Valor Venda']]

# Colunas

tabela_analise = tabela1.drop(columns='Data_Base') # Exclui a coluna  

display(tabela_analise)

# Valores vazio 

tabela_analise['Loja']= tabela_analise['Loja'].fillna('Online') # Substitui os valores vazios por 'Desconecido'

#  tabela_analise = tabela_analise.dropna() --> Exclui as linhas com valores vazios

tabela_analise['Loja'] = tabela_analise['Loja'].fillna(0) # Substitui os valores vazios por 0

# Tipos de dados

tabela_analise['Data'] = pd.to_datetime(tabela_analise['Data'], format="%Y-%m-%d") # Converte a coluna 'Data' para o tipo datetime

# Padronização de dados

tabela_analise['Loja'] = tabela_analise['Loja'].str.upper() # Converte os valores da coluna 'Loja' para maiúsculo
tabela_analise['Loja'] = tabela_analise['Loja'].str.strip() # Remove os espaços em branco da coluna 'Loja'  
tabela_analise['Loja'] = tabela_analise['Loja'].str.replace(' ', '_') # Substitui os espaços por underline na coluna 'Loja'
tabela_analise['Loja'] = tabela_analise['Loja'].str.title() # Substitui os traços por underline na coluna 'Loja'

# Duplicados

tabela_analise = tabela_analise.drop_duplicates(subset=['ID_Pedido', 'Loja'])  # Exclui as linhas duplicadas

display(tabela_analise)
```

```python
#analisar --> filtrar

tabela_analise = tabela_analise.sort_values(by='Data')
tabela_analise = tabela_analise.reset_index(drop=True)

#.loc --> Nome coluna/ NOme linha 
id_pedido = 4

loja = tabela_analise.loc[3, "Loja"]
produto = tabela_analise.loc[3, "Produto"] 
cliente = tabela_analise.loc[3, "Cliente"]

print(loja, produto, cliente)
#.iloc ---> Por posição 
loja = tabela_analise.iloc[3, 2]
produto = tabela_analise.iloc[3, 3]
cliente = tabela_analise.iloc[3, 6]

print(loja, produto, cliente)
# condicional

# exportar pedaços da base

#tabela_vendas_sp = tabela_analise

display(tabela_analise)
```
