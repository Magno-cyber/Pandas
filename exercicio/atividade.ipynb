```python
import pandas as pd
import io

csv_data = """match_id,agent,kills,deaths,assists,combat_score,outcome
1,Sage,15,10,8,210,Win
2,K.O,12,14,10,180,Loss
3,sage,NaN,8,12,150,win
4,Jett,25,12,4,310,Win
5,KAY/O,8,15,9,120,LOSS
6,Sage,20,5,15,250,Win
7,K.O,-5,10,7,90,Loss
8,Reyna,30,15,2,350,win
9,sage,10,10,10,185,
10,KAY/O,14,12,NaN,195,Win
"""

# Carregando os dados para o DataFrame
df = pd.read_csv(io.StringIO(csv_data), low_memory=False)
```

```python
# Tratamento de dados

# 1. Padronização de Texto (Strings)
# O dicionário no replace faz múltiplas substituições de uma vez só
df['agent'] = df['agent'].replace({'sage': 'Sage', 'K.O': 'KAY/O'})

# .str.capitalize() transforma tudo para formato de título ("Win", "Loss")
df['outcome'] = df['outcome'].str.capitalize()


# 2. Tratamento de Valores Nulos (NaN)
df['kills'] = df['kills'].fillna(df['kills'].median())
df['assists'] = df['assists'].fillna(df['assists'].median())
df['outcome'] = df['outcome'].fillna('Unknown')


# 3. Correção de Inconsistências Lógicas
# .abs() converte números negativos em valores absolutos (positivos)
df['kills'] = df['kills'].abs()


# 4. Criação de Novas Colunas (Feature Engineering)
df['KDA_Ratio'] = ((df['kills'] + df['assists']) / df['deaths']).round(2)


# 5. Filtragem (Opcional)
# Usando parênteses para separar as duas condições com o operador "E" (&)
df_suporte = df[(df['agent'].isin(['Sage', 'KAY/O'])) & (df['outcome'] == 'Win')]


display(df)
```
