```python
import pandas as pd
import io

csv_data = """match_id,agent,kills,deaths,assists,combat_score,outcome
1,Sage,15,10,8,210,Win
2,K.O,12,14,10,180,Loss
3,sage,NaN,8,12,150,win
4,Jett,25,12,4,310,Win
5,KAY/O,8,15,9,120,LOSS
6,Sage,20,5,15,250,Win
7,K.O,-5,10,7,90,Loss
8,Reyna,30,15,2,350,win
9,sage,10,10,10,185,
10,KAY/O,14,12,NaN,195,Win
"""

# Carregando os dados para o DataFrame
df = pd.read_csv(io.StringIO(csv_data), low_memory=False)
```

```python
# Tratamento de dados

df_analise = df.copy() # Cria uma cópia do DataFrame df para o DataFrame df_analise

# NUlos 

df_analise['assists'] = df_analise['assists'].fillna(0) # Usado para preencher valores ausentes na coluna "assists" do DataFrame df_analise com 0
df_analise['kills'] = df_analise['kills'].fillna(0) # Usado para preencher valores ausentes na coluna "kills" do DataFrame df_analise com a média da coluna "kills"
df_analise['outcome'] = df_analise['outcome'].fillna('Unknown') # Usado para preencher valores ausentes na coluna "outcome" do DataFrame df_analise com 'Unknown'

# Padronização de dados

df_analise['agent'] = df_analise['agent'].str.strip()


display(df_analise) # Usado para exibir o DataFrame df_analise
```
